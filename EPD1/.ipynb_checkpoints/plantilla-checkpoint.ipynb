### Solución Ejercicio 03

```python
a = np.random.rand(60,60)
b = np.random.rand(60,60)
print(a.shape)
```

```python
suma_opt1 = np.zeros((60,60))
suma_opt11 = np.zeros((60,60))
```

```python
start_time_opt1 = time.time()

for i in range(len(a)):
    for j in range(len(a[i])):
        suma_opt1[i][j] = a[i][j]+b[i][j]
        suma_opt11[i,j] = a[i,j]+b[i,j]

print("EJ3: SUM option1: shape: ", suma_opt1.shape, " , total sum: ", suma_opt1.sum(), ", time: ", time.time()-start_time_opt1)
print("EJ3: SUM option11: shape: ", suma_opt11.shape, " , total sum: ", suma_opt1.sum(), ", time: ", time.time()-start_time_opt1)
```

```python
suma_opt2 = np.zeros((60,60))
start_time_opt2 = time.time()

suma_opt2 = a+b #
print("EJ3: SUM option2: shape: ", suma_opt2.shape, " , total sum: ", suma_opt2.sum(), ", time: ", time.time()-start_time_opt2)
```

```python
start_time_opt3 = time.time()
suma_opt3 = np.sum([a,b])      # SUMA TODO

print("EJ3: SUM option3: shape: ", suma_opt3.shape, " , total sum: ", suma_opt3.sum(), ", time: ", time.time()-start_time_opt3)
```

### Solución Ejercicio 04

```python
# Escribe tu solución aquí
a4 = np.random.rand(10,10)
print(a4.shape)
```

```python
# USANDO DOS BUCLES

# Asignamos un bloque de memoria para una nueva matriz 10x10
mult_opt1 = np.empty_like(a4)
# Bucle que recorre los digitos de cada indice y lo multiplica por su numero de fila (i)
for i in range(len(a4)):
    for j in range(len(a4[i])):
        mult_opt1[i,j] = a4[i][j] * i

# print("Matriz resultante: ")
# print(mult_opt1)

# Creamos un DataFrame para que se vea bonito
pd.DataFrame(mult_opt1)
```

```python
# USANDO UN BUCLE
mult_opt2 = np.empty_like(a4)

# Recorremos la filas a traves de sus indices  y lo multiplicamos por su numero de fila
for i in range(a4.shape[0]):
    mult_opt2[i] = a4[i] * i

pd.DataFrame(mult_opt2)
```

### Solución Ejercicio 05

```python
# Escribe tu solución aquí
a5 = np.random.rand(10,10)
b5 = np.random.rand(10,10)

print(a5.shape)
print(b5.shape)
```

```python
# Usando dos bucles
mult_EJ5_opt1 = np.empty_like(a5)

for i in range(len(a5)):
    for j in range(len(a5[i])):
        mult_EJ5_opt1[i,j] = a5[i,j]*b5[i,j]

pd.DataFrame(mult_EJ5_opt1)
```

```python
# USANDO UN BUCLE
mult_EJ5_opt2 = np.empty_like(a5)

for i in range(a5.shape[0]):
    mult_EJ5_opt2[i] = a5[i] * b5[i]

pd.DataFrame(mult_EJ5_opt2)
```

### Solución Ejercicio 07

```python
# Escribe tu solución aquí
A = np.array([
            [8,3,-2],
            [-4,7,5],
            [3,4,-12]])
B = np.array([9,15,35])

## METODO np.linalg.inv()
eq_res_opt1 = np.linalg.inv(A).dot(B)
eq_res_opt1_2 = np.dot(np.linalg.inv(A),B)

print("\nEJ7: Sol.1: ", eq_res_opt1)
print("\nEJ7: Sol.1_2: ", eq_res_opt1_2)

## METODO np.linalg,solve(A,b)
eq_res_opt2 = np.linalg.solve(A,B)
print("\nEJ7: Sol.2: ", eq_res_opt2)
```
